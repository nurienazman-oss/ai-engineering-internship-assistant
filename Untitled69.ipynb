```python
# Cell 4 — REQUIRED: Source Selection + Lineage Utilities
# Run ONCE after Cell 1 and Cell 3

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

# Select the exact source before parsing.
# Never automatically take the first file by extension.
def select_source(relative_path=None, filename=None, suffixes=None):
  if relative_path:
    selected = RAW_DIR / relative_path
    if not selected.is_file():
        raise FileNotFoundError(
          f"Source not found: {selected}"
        )
    return selected

  matches = sorted(
    p for p in RAW_DIR.rglob("*")
    if p.is_file()
    and (filename is None or p.name == filename)
    and (suffixes is None or p.suffix.lower() in suffixes)
  )

  print(
    "Candidates:",
    [str(p.relative_to(RAW_DIR)) for p in matches]
  )
  if len(matches) != 1:
    raise ValueError(
      "Select exactly one source. "
      "Set relative_path to an exact file under RAW_DIR."
    )

  return matches[0]

def source_hash(path):
  return hashlib.sha256(
    Path(path).read_bytes()
  ).hexdigest()

def save_manifest(records):
  manifest = {
    "pipeline_version": "week3-v1",
    "generated_at_utc": datetime.now(timezone.utc).isoformat(),
    "records": records
  }

  target = BASE_DIR / "data" / "week3_manifest.json"
  target.parent.mkdir(parents=True, exist_ok=True)
  target.write_text (
    json.dumps(
      manifest,
      indent=2,
      ensure_ascii=False
    ),
    encoding="utf-8"
  )
  print("Saved lineage:", target)
  return manifest

# Confirmation messages
print("Cell 4 completed successfully.")
print("Source selection and lineage utilities are ready.")
print("Available functions:")
print("- select_source()")
print("- source_hash()")
print("- save_manifest()")
```

```python
# CELL 4A — COMMON LINEAGE REGISTRY
# REQUIRED: Run once before selected Part B pathways
from pathlib import Path
lineage_entries = {}

def register_lineage(source, output, processing):
  source = Path(source).resolve()
  output = Path(output).resolve()
  # Validate source
  if not source.is_file():
    raise FileNotFoundError(
      f"Source file not found: {source}"
    )
  # Validate prepared output
  if not output.is_file():
    raise FileNotFoundError(
      f"Prepared output not found: {output}"
    )
  if output.stat().st_size == 0:
    raise ValueError(
      f"Prepared output is empty: {output}"
    )
  # Validate processing description
  if not isinstance(processing, str) or not processing.strip():
    raise ValueError(
      "A processing description is required."
    )
  processing = processing.strip()
  if processing.upper().startswith("REPLACE"):
    raise ValueError(
      "Replace the placeholder with the actual method."
    )
  # Ensure source and output belong to project workspace
  source.relative_to(RAW_DIR.resolve())
  output.relative_to(PROCESSED_DIR.resolve())
  # Same source/output pair is updated, not duplicated
  key = (str(source), str(output))
  lineage_entries[key] = {
    "source": source,
    "output": output,
    "processing": processing
  }
  print(
    "Registered:",
    source.name,
    "->",
    output.name
  )

print("Common lineage registry ready.")
print("Registered entries:", len(lineage_entries))
```

```python
import re, json

# ---------- Clause splitter ----------
# (?<![\d.]) stops numbers like 2.7.1 from being cut in the middle
CLAUSE = re.compile(r"(?=(?:SEKSYEN\s+\d+\s*:|(?<![\d.])\d+\.\d+(?:\.\d+)*\s+[A-Z(]))")

def split_blocks(text):
    text = re.sub(r"[ \t]+", " ", text).strip()
    if text.count("\n") < 5:          # S01: almost no line breaks
        return [b.strip() for b in CLAUSE.split(text) if b.strip()]
    return [b.strip() for b in re.split(r"\n\s*\n", text) if b.strip()]

def fixed_chunks(s, size=500):
    return [s[i:i + size] for i in range(0, len(s), size)]

def page_section_chunks(pages, size=600):
    out, section = [], None
    for page_no, content in pages:
        cur, cur_sec = "", section
        for b in split_blocks(content):
            h = re.match(r"^(SEKSYEN\s+\d+\s*:[^\d]{3,60}|\d+\.\d+(?:\.\d+)*)", b)
            if h:
                section = h.group(1).strip()
            if len(b) > size:
                if cur: out.append((page_no, cur_sec, cur)); cur = ""
                out.extend((page_no, section, b[i:i + size]) for i in range(0, len(b), size))
                cur_sec = section
            elif len(cur) + len(b) + 1 <= size:
                if not cur: cur_sec = section
                cur += (" " if cur else "") + b
            else:
                out.append((page_no, cur_sec, cur)); cur = b; cur_sec = section
        if cur: out.append((page_no, cur_sec, cur))
    return out

# ---------- B3: compare two configurations ----------
KEY_START, KEY_END = "Memperoleh kredit dapat sebanyak 70%", "standard program akademik"
tot_fixed = tot_psc = 0
for sid, meta in SOURCES.items():
    f = select_source(relative_path=meta["file"])
    pages = read_pages(f)
    clean = re.sub(r"[ \t]+", " ", "\n".join(t for _, t in pages)).strip()
    fx  = fixed_chunks(clean, 500)
    psc = page_section_chunks(pages, 600)
    no_sec = sum(1 for _, s, _ in psc if not s)
    tiny   = sum(1 for _, _, t in psc if len(t) < 80)
    tot_fixed += len(fx); tot_psc += len(psc)
    print(f"{sid}: fixed500={len(fx)} | page/section600={len(psc)} | no section={no_sec} | tiny(<80)={tiny}")

# Key-sentence test (the 70% credit rule, only in S01)
f1 = select_source(relative_path=SOURCES["S01"]["file"])
p1 = read_pages(f1)
c1 = re.sub(r"[ \t]+", " ", "\n".join(t for _, t in p1)).strip()
print("\nKey sentence intact in one chunk?")
print(" fixed500        :", any(KEY_START in c and KEY_END in c for c in fixed_chunks(c1, 500)))
hit = [(p, s) for p, s, t in page_section_chunks(p1, 600) if KEY_START in t and KEY_END in t]
print(" page/section600 :", bool(hit), "| (page, section):", hit)
print("Total chunks -> fixed:", tot_fixed, "| page/section:", tot_psc)
```

```python
records, chosen = [], "page-section-aware-char-600"
for sid, meta in SOURCES.items():
    f = select_source(relative_path=meta["file"])
    sha = source_hash(f)
    for i, (page, section, text) in enumerate(page_section_chunks(read_pages(f), 600), 1):
        if section is None and meta["doc_type"] == "form":
            section = "(form, no section numbers)"
        records.append({
            "record_id": f"{sid}-{f.stem[:30]}-{i:04d}",
            "text": text,
            "source_id": sid,
            "source": str(f.relative_to(RAW_DIR)),
            "source_sha256": sha,
            "page": page,                 # PDF page, not the printed page number
            "section": section,
            "version": meta["version"],
            "status": meta["status"],
            "doc_type": meta["doc_type"],
            "processing": chosen,
        })

prepared_path = PROCESSED_DIR / "prepared_text.jsonl"
with prepared_path.open("w", encoding="utf-8") as out:
    for r in records:
        out.write(json.dumps(r, ensure_ascii=False) + "\n")

# Basic validation
assert prepared_path.stat().st_size > 0
assert records and all(r["text"].strip() for r in records)
with prepared_path.open(encoding="utf-8") as fh:
    assert len([1 for line in fh if line.strip()]) == len(records)
print("Chunks:", len(records),
      "| with page:", sum(r["page"] is not None for r in records),
      "| with section:", sum(bool(r["section"]) for r in records))
print(records[0])
```
